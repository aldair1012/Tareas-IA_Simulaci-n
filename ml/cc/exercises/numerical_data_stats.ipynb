```python
# @title Setup - Install relevant modules
!pip install matplotlib~=3.10.0 \
  pandas~=2.2.0

# @title Setup - Import relevant modules
import pandas as pd
from matplotlib import pyplot as plt
import io

# Ajustar la granularidad de los informes.
pd.options.display.max_rows = 10
pd.options.display.float_format = "{:.1f}".format

# Cargar el nuevo dataset
training_df = pd.read_csv("https://raw.githubusercontent.com/aldair1012/Tareas-IA_Simulaci-n/aldair1012-rama1/IA/player-data-full-2025-june.csv", low_memory=False, encoding='latin1')

print("Dataset cargado. Aquí están las primeras filas y un resumen de la información:")
display(training_df.head())
display(training_df.info())

print("\nEstadísticas básicas del dataset:")
display(training_df.describe())

# @title Define las funciones de trazado

def plot_the_dataset(feature, label, number_of_points_to_plot):
  """Traza N puntos aleatorios del dataset."""

  # Etiquetar los ejes.
  plt.xlabel(feature)
  plt.ylabel(label)

  # Crear un diagrama de dispersión a partir de N puntos aleatorios del dataset.
  random_examples = training_df.sample(n=number_of_points_to_plot)
  plt.scatter(random_examples[feature], random_examples[label])

  # Renderizar el diagrama de dispersión.
  plt.show()

def plot_a_contiguous_portion_of_dataset(feature, label, start, end):
  """Traza los puntos de datos desde el inicio hasta el final."""

  # Etiquetar los ejes.
  plt.xlabel(feature + " (índice)") # Modificado para reflejar que 'contiguous' es por índice aquí
  plt.ylabel(label)

  # Crear un diagrama de dispersión.
  plt.scatter(training_df[feature][start:end], training_df[label][start:end])

  # Renderizar el diagrama de dispersión.
  plt.show()

print("\nFunciones de trazado definidas: plot_the_dataset y plot_a_contiguous_portion_of_dataset.")

feature_column = 'height_cm'
label_column = 'overall_rating'

print(f"\nVisualizando 50 puntos aleatorios para: '{feature_column}' vs '{label_column}'")
plot_the_dataset(feature_column, label_column, number_of_points_to_plot=50)

print(f"\nVisualizando 500 puntos aleatorios para: '{feature_column}' vs '{label_column}'")
plot_the_dataset(feature_column, label_column, number_of_points_to_plot=500)

print(f"\nVisualizando una porción contigua (primeros 100 jugadores) para: '{feature_column}' vs '{label_column}'")
plot_a_contiguous_portion_of_dataset(feature_column, label_column, start=0, end=100)

print("\nLos análisis de 'estadísticas por semana' y 'visualización por día' del Colab original no se han incluido directamente aquí porque no son aplicables a la estructura de este nuevo dataset. Deberías definir tus propios rangos y criterios de segmentación si deseas realizar un análisis similar con los datos de jugadores.")
```
